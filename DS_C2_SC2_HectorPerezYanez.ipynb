### a. Extraer la información del archivo.

```python
import pandas as pd

country_vaccinations = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/bases de datos/country_vaccinations.csv')

country_vaccinations.head()
```

### b. Mostrar la estructura y tipos de datos de cada columna, asegurando que las fechas sean datetime64

```python
country_vaccinations['date'] = pd.to_datetime(country_vaccinations['date'])
country_vaccinations.info()
```

### c. Determinar la cantidad de vacunas aplicadas de cada compañía (agrupando por la columna vaccines y realizando la sumatoria)

```python
pd.set_option('display.max_rows', 100)
pd.options.display.float_format = '{:,.0f}'.format

vacunas_por_compania = (country_vaccinations
                        .groupby('vaccines')['daily_vaccinations']
                        .sum()
                        .sort_values(ascending=False)
                        .reset_index())
vacunas_por_compania
```

### d. Obtener la cantidad de vacunas aplicadas en todo el mundo

```python
total_mundial = country_vaccinations['daily_vaccinations'].sum()
print(f'Vacunas aplicadas en todo el mundo: {total_mundial:,.0f}')
```

### e. Calcular el promedio de vacunas aplicadas por país

```python
promedio_por_pais = (country_vaccinations
                     .groupby('country')['daily_vaccinations']
                     .mean()
                     .sort_values(ascending=False)
                     .reset_index())
promedio_por_pais
```

### f. Determinar la cantidad de vacunas aplicadas el día 29/01/21 en todo el mundo

```python
vacunas_29_ene = country_vaccinations.loc[
    country_vaccinations['date'] == '2021-01-29', 'daily_vaccinations'
].sum()
print(f'Vacunas aplicadas el 29/01/2021 en todo el mundo: {vacunas_29_ene:,.0f}')
```

### g. Crear un dataframe nuevo denominado conDiferencias con los datos originales y una columna derivada (diferencias) con las diferencias entre daily_vaccinations y daily_vaccinations_raw

```python
conDiferencias = country_vaccinations.copy()
conDiferencias['diferencias'] = conDiferencias['daily_vaccinations'] - conDiferencias['daily_vaccinations_raw']

columnas = ['country', 'date', 'daily_vaccinations', 'daily_vaccinations_raw', 'diferencias']
conDiferencias.dropna(subset=['diferencias'])[columnas].head(10)
```

### h. Obtener el periodo de tiempo entre el registro con fecha más reciente y el registro con fecha más antigua

```python
fecha_min = country_vaccinations['date'].min()
fecha_max = country_vaccinations['date'].max()
periodo = fecha_max - fecha_min

print(f'Fecha más antigua: {fecha_min.date()}')
print(f'Fecha más reciente: {fecha_max.date()}')
print(f'Periodo de tiempo: {periodo}')
```

### i. Crear un dataframe nuevo denominado conCantidad con los datos originales y una columna derivada (canVac) con la cantidad de vacunas utilizadas cada día (separando la columna vaccines por el carácter ,)

```python
conCantidad = country_vaccinations.copy()
conCantidad['canVac'] = conCantidad['vaccines'].str.split(',').str.len()

conCantidad[['country', 'date', 'vaccines', 'canVac']].head(10)
```

### j. Generar un dataframe denominado antes20 con todos los registros que se hayan realizado antes del 20 de diciembre de 2020

```python
antes20 = country_vaccinations[country_vaccinations['date'] < '2020-12-20']

print(f'Registros antes del 20/12/2020: {len(antes20)}')
antes20
```

### k. Obtener un dataframe denominado pfizer con todos los registros donde se haya utilizado la vacuna Pfizer

```python
pfizer = country_vaccinations[country_vaccinations['vaccines'].str.contains('Pfizer')]

print(f'Registros con vacuna Pfizer: {len(pfizer)}')

pfizer[['country', 'date', 'vaccines']].head(10)
```

### l. Almacenar los dataframes conDiferencias, conCantidad, antes20 y pfizer en un archivo de Excel denominado resultadosReto.xlsx, cada uno en una hoja diferente

```python
ruta_salida = '/content/drive/MyDrive/Colab Notebooks/bases de datos/resultadosReto.xlsx'

with pd.ExcelWriter(ruta_salida) as writer:
    conDiferencias.to_excel(writer, sheet_name='conDiferencias', index=False)
    conCantidad.to_excel(writer, sheet_name='conCantidad', index=False)
    antes20.to_excel(writer, sheet_name='antes20', index=False)
    pfizer.to_excel(writer, sheet_name='pfizer', index=False)

print(f'Archivo guardado en: {ruta_salida}')
```
